### Step 1 & 2: Clone the Repository

First, make sure to enable a GPU/High-RAM Runtime if needed. Then, clone the repository from GitHub:

```python
!git clone https://github.com/tangency-labs/solidcraft.git
%cd solidcraft
```

### Step 3: Install the Rust Toolchain

SolidCraft is written in Rust, so we need to install `rustc` and `cargo`:

```python
%%bash
curl --proto '=https' --tlsv1.2 -sSf https://sh.rustup.rs | sh -s -- -y
source "$HOME/.cargo/env"
rustc --version
cargo --version
```

### Step 4: Build the Workspace

Build the workspace in release mode for maximum performance:

```python
%%bash
if [ ! -d "/content/solidcraft" ]; then
  echo "Repository not found. Re-cloning..."
  git clone https://github.com/tangency-labs/solidcraft.git /content/solidcraft
fi
cd /content/solidcraft

# Source rustup toolchain paths
export PATH="$HOME/.cargo/bin:$PATH"

# Locate the modern cargo executable
CARGO_BIN=$(which cargo)

# Verify version
$CARGO_BIN --version

# Run build using the modern toolchain
$CARGO_BIN build --release
```

### Step 5: Run Examples or Use the Headless CLI / MCP Server

Run the CLI to generate models or start the MCP server:

```python
%%bash
source "$HOME/.cargo/env"
# Example of running a sample command or building reference parts via CLI
cargo run --release -p solidcraft-cli -- --help

# Or to launch the MCP server component:
# cargo run --release -p solidcraft-cli -- mcp
```

### Step 6: Explore Available Commands

Now that the CLI is built, let's check the available CAD commands supported by SolidCraft.

```python
%%bash
source "$HOME/.cargo/env"
# List all available commands in SolidCraft
cargo run --release -p solidcraft-cli -- commands
```

```python
%%bash
# If repository is not found in /content, clone it again
if [ ! -d "/content/solidcraft" ]; then
  echo "Cloning solidcraft repository to /content/solidcraft..."
  git clone https://github.com/tangency-labs/solidcraft.git /content/solidcraft
fi

cd /content/solidcraft

# Source cargo path to make sure the binary environment is loaded
source "$HOME/.cargo/env"

# Run the SolidCraft CLI with the --help command
cargo run --release -p solidcraft-cli -- --help
```
